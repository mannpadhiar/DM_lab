### Step 4: Define Apriori Function
This function finds frequent itemsets of size 1, 2, and 3 with minimum support.

```python
from itertools import combinations

def findPri(Tdata,min_support):
    n = len(Tdata)
    
    result = []

    for k in [1,2,3]:
        for i in combinations(Tdata.columns,k):
            mask = Tdata[list(i)].all(axis=1)
            support = mask.sum()/n


            if support >= min_support:
                print(f"{frozenset(i)} -> {round(support,2)}")
                result.append((frozenset(i),round(support,2)))
    return result


```

### Step 5: Run Apriori
Set `min_support = 0.6` and display the frequent itemsets.

```python
findPri(Tdata,.6)
```

### Step 6 Display as a DataFrame

```python
result_df = pd.DataFrame(findPri(Tdata,.6),columns= ['Itemset','Support'])
result_df
```
